# OME-Zarr multichannel viewer (anywidget)

This notebook builds a small two-channel fluorescence dataset
(`skimage.data.cells3d` — cell membranes + nuclei) and opens it in oz-viewer's
**anywidget** viewer, which renders **inline** in the notebook (no desktop
window).

The channel axis is detected from the OME-Zarr metadata, so the viewer opens in
multichannel mode with a per-channel control panel.

**Controls**
- **Left dock** — per-channel colormap, contrast limits, opacity, visibility.
- **Bottom dock** — toggle 2D / 3D.
- **2D mode** — pan (left-drag), zoom (scroll), z-slice slider below the canvas.
- **3D mode** — orbit (left-drag), zoom (scroll), pan (right-drag).

**Setup.** From the repo root, install the example extras into the environment
(the anywidget canvas backend currently needs `rendercanvas` from git main):

```bash
VIRTUAL_ENV=.venv uv pip install -e '.[examples]'
VIRTUAL_ENV=.venv uv pip install "rendercanvas @ git+https://github.com/pygfx/rendercanvas.git"
```


````python
from oz_viewer.data import make_cells3d_zarr
from oz_viewer.viewer import display_viewer
````

````python
# Writes cells3d.ome.zarr next to this notebook (skipped if it already exists).
# The first run downloads the sample data via pooch, so it needs network access.
path = make_cells3d_zarr()
uri = f"file://{path}"
uri
````

`display_viewer` extracts the OME-Zarr geometry, builds the anywidget viewer —
auto-detecting the channel axis, so the two channels open in multichannel mode —
and renders it inline. It is non-blocking; keep the returned handle to tear the
viewer down (`handle.close()`). `min_canvas_size` sets the minimum canvas size in pixels
 `(width, height)` the canvas is allowed to shrink to as the notebook /
dock is resized.

````python
handle = display_viewer(uri, sidecar=True, min_canvas_size=(100, 300))
````
